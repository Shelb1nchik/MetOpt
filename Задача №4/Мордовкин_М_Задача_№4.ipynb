```python
# Данные банков: 17 значений — по 17 критериям
# Порядок значений соответствует критериям 1..17 из условия.

banks = {
    'ВТБ24':              [9.7,  13.2, 0,   1.5, 1.3, 5,  50, 14, 25,   250,   21, 65, 1,   4,  3500,  0.1,  11.45],
    'Санкт-Петербург':    [10.9, 14,   10,  1.2, 1,   1,  25, 5,  37.5, 537.5, 18, 65, 1,   6,  1500,  0.04, 12.45],
    'Райффазен':          [10,   11.5, 10,  1,   1,   1,  25, 5,  25,   375,   21, 60, 1,   3,  21200, 0.9,  10.75],
    'МДМ':                [10.7, 12.5, 10,  1.5, 1.2, 3,  25, 4,  12.5, 75,    21, 65, 1,   4,  2400,  0.5,  11.6],
    'Сбербанк':           [9.5,  12,   10,  0.9, 0,   1,  30, 18, 25,   15,    18, 75, 0.5, 6,  0,     0.75, 10.75],
    'GeMoney':            [4,    7.05, 5,   0.7, 1,   1,  30, 3,  37.5, 500,   21, 69, 3,   6,  10000, 2,    5.525],
    'DeltaCredit':        [8,    11,   15,  0.56,1,   7,  25, 25, 3,    75,    18, 55, 1,   6,  3500,  1,    9.5],
    'Абсолют':            [13,   15,   10,  0.9, 1,   3,  25, 5,  25,   250,   21, 60, 2,   24, 10500, 0.3,  11],
    'ПСБ':                [9,    11,   30,  0.7, 0,   7,  25, 1,  7.5,  500,   21, 70, 1,   3,  3000,  0.5,  10],
    'КМБ':                [9,    11.5, 15,  0.9, 1,   3,  30, 5,  12.5, 75,    23, 60, 0.25,3,  0,     0.16, 10.25],
    'ОТП':                [9.5,  11.8, 10,  0.8, 0.8, 1,  30, 5,  50,   250,   21, 70, 2,   4,  0,     0.2,  10.65],
    'ЮниКредит':          [9,    11.9, 15,  0.7, 1.5, 1,  20, 7,  37.5, 250,   21, 60, 2,   6,  0,     0.2,  10.45],
    'Городской ипотечный':[9,    11.5, 15,  0.4, 0.8, 1,  25, 3,  50,   250,   21, 54, 1,   4,  2800,  3,    10.25],
    'РосБанк':            [9.5,  11.5, 10,  1.5, 0,   1,  25, 5,  12.5, 187.5, 20, 60, 0.5, 6,  4200,  0.5,  10.5],
    'Северная казна':     [8.75, 10.5, 25,  1,   3,   10, 25, 5,  7.5,  300,   21, 60, 1.5, 18, 0,     0.5,  9.625],
}

# Порядок критериев (совпадает с порядком значений в списках выше)
criteria = [
    'Мин. ставка, %', 'Макс. ставка, %', 'Первый взнос, %',
    'Страхование, %', 'Комиссия, %', 'Мин. срок, лет',
    'Макс. срок, лет', 'Рассмотрение, дн.', 'Макс. сумма, млн',
    'Мин. сумма, тыс.', 'Мин. возраст, лет', 'Макс. возраст, лет',
    'Мин. общий стаж, лет', 'Мин. текущий стаж, мес.',
    'Расходы, руб.', 'Плата за просрочку, %', 'Средняя ставка, %',
]

# Направление критерия: 'min' — меньше лучше, 'max' — больше лучше
directions = ['min','min','min','min','min','min','max','min',
              'max','min','min','max','min','min','min','min','min']
```

```python
# Часть 1. Задаём тест для каждого критерия и упорядочиваем
# банки по числу пройденных тестов.
# Тест — пороговое условие. Если значение банка его проходит —
# банк получает 1 балл. Побеждает тот, у кого больше баллов.

def make_tests():
    """Возвращает список тестов: (индекс критерия, направление, порог).

    Каждый порог выбран так, чтобы отсеять банки с заведомо
    невыгодными или нетипичными условиями, оставив «приемлемые».
    """
    return [
        # 1. Мин. ставка <= 10 %
        #    Средняя минимальная ставка по таблице ≈ 9.5 %.
        #    10 % — уровень «не хуже среднего по рынку».
        (0,  'min', 10.0),

        # 2. Макс. ставка <= 12 %
        #    Максимальная ставка 12 % — типичный верхний предел,
        #    выше которого кредит считается дорогим.
        (1,  'min', 12.0),

        # 3. Первый взнос <= 10 %
        #    Стандартное требование рынка: 10–20 %.
        #    10 % — минимальный «льготный» порог.
        (2,  'min', 10.0),

        # 4. Ежемесячное страхование <= 1 %
        #    Большинство банков держат страхование в диапазоне 0.4–1.5 %.
        #    1 % — медианный уровень.
        (3,  'min', 1.0),

        # 5. Комиссия банка <= 1 %
        #    Комиссии выше 1 % — уже заметная переплата.
        (4,  'min', 1.0),

        # 6. Мин. срок кредита <= 3 лет
        #    Чем меньше минимальный срок, тем гибче условия для заёмщика.
        #    3 года — уровень, при котором банк не навязывает длинный кредит.
        (5,  'min', 3.0),

        # 7. Макс. срок кредита >= 25 лет
        #    25 лет — типовой максимум по ипотеке в РФ.
        (6,  'max', 25.0),

        # 8. Рассмотрение заявки <= 7 дней
        #    Большинство банков рассматривают заявку за 3–7 дней.
        #    7 — верхняя граница «быстрого» решения.
        (7,  'min', 7.0),

        # 9. Макс. сумма кредита >= 30 млн руб.
        #    30 млн — уровень, достаточный для покупки жилья в крупном городе.
        (8,  'max', 30.0),

        # 10. Мин. сумма кредита <= 300 тыс. руб.
        #    Чем ниже минимальная сумма, тем доступнее кредит.
        #    300 тыс. — низкий порог входа.
        (9,  'min', 300.0),

        # 11. Мин. возраст заёмщика <= 21 года
        #    21 год — стандартный минимум у большинства банков.
        (10, 'min', 21.0),

        # 12. Макс. возраст заёмщика >= 65 лет
        #    65 лет — типовой максимум на момент погашения кредита.
        (11, 'max', 65.0),

        # 13. Мин. общий стаж <= 1 года
        #    Банки обычно требуют от 1 года общего стажа.
        (12, 'min', 1.0),

        # 14. Мин. текущий стаж <= 6 мес.
        #    6 месяцев на текущем месте — минимальное требование рынка.
        (13, 'min', 6.0),

        # 15. Расходы на оформление <= 3000 руб.
        #    3000 руб. — умеренная сумма; выше — уже заметные расходы.
        (14, 'min', 3000.0),

        # 16. Плата за просрочку <= 0.5 %
        #    Меньше — мягче условия для заёмщика.
        #    0.5 % — медианное значение по таблице.
        (15, 'min', 0.5),

        # 17. Средняя ставка <= 11 %
        #    11 % — уровень «не хуже среднего» по рынку ипотеки.
        (16, 'min', 11.0),
    ]


def passes(value, direction, threshold):
    """Проверяет, проходит ли значение тест."""
    if direction == 'min':
        return value <= threshold
    else:
        return value >= threshold


def count_tests(bank_values, tests):
    """Считает число пройденных тестов."""
    return sum(1 for (idx, d, th) in tests if passes(bank_values[idx], d, th))


tests = make_tests()

# Считаем баллы для каждого банка
results = []
for name, vals in banks.items():
    score = count_tests(vals, tests)
    results.append((name, score))

# Сортируем по убыванию балла; при равенстве — по имени (для устойчивости)
results.sort(key=lambda t: (-t[1], t[0]))

print("Часть 1. Упорядочение по числу пройденных тестов")
for i, (name, score) in enumerate(results, 1):
    print(f"{i:2d}. {name:22s} — пройдено {score} / {len(tests)}")
```

```python
# Часть 2. Парето-оптимальность
# Банк A доминирует банк B, если по всем критериям A не хуже B
# и хотя бы по одному — строго лучше.
# Банки, которых никто не доминирует, образуют Парето-фронт.

def to_scores(values):
    """Переводим значения в 'очки', где больше = лучше.
    Для критериев min берём -value (чтобы меньшие значения давали больший балл)."""
    return tuple(-v if directions[i] == 'min' else v
                 for i, v in enumerate(values))


# Заранее считаем очки для каждого банка
scores = {name: to_scores(v) for name, v in banks.items()}


def dominates(a, b):
    """Возвращает True, если банк a доминирует банк b."""
    sa = scores[a]
    sb = scores[b]
    not_worse_all = all(x >= y for x, y in zip(sa, sb))
    strictly_better_one = any(x > y for x, y in zip(sa, sb))
    return not_worse_all and strictly_better_one


# Для каждого банка собираем список тех, кто его доминирует
dominators = {name: [] for name in banks}
for name in banks:
    for other in banks:
        if other != name and dominates(other, name):
            dominators[name].append(other)

# Парето-оптимальные — у кого список доминаторов пуст
pareto = [name for name in banks if not dominators[name]]
dominated = [name for name in banks if dominators[name]]

print("Часть 2. Парето-оптимальные банки")
for i, name in enumerate(pareto, 1):
    print(f"{i}. {name}")

print("\nНЕ Парето-оптимальные банки (доминируются):")
for name in dominated:
    print(f"\n '{name}' доминируется банками:")
    for d in dominators[name]:
        print(f"   • {d}")
```

```python
# Часть 3. Лексикографическое упорядочение
# 1) Задаём порядок критериев по важности (от самого важного).
# 2) Сортируем банки лексикографически: сравниваем по первому
#    критерию, при равенстве — по второму и т.д.
# 3) Повторяем с обратным порядком критериев.

# Порядок критериев по важности (индексы в массиве criteria)
importance = [16, 0, 1, 2, 8, 6, 14, 3, 4, 7, 9, 10, 11, 5, 12, 13, 15]

print("Порядок критериев по важности:")
for pos, idx in enumerate(importance, 1):
    print(f"  {pos:2d}. {criteria[idx]} ({directions[idx]})")


def lex_key(values, order):
    """Ключ для лексикографической сортировки.
    Переводим значение в 'очки' (больше = лучше) и берём с минусом,
    чтобы sorted() по возрастанию дал нужный порядок."""
    key = []
    for idx in order:
        v = values[idx]
        score = -v if directions[idx] == 'min' else v
        key.append(-score)   # инвертируем: меньшее значение ключа = лучше
    return tuple(key)


#Прямой порядок критериев
order_forward = importance
banks_forward = sorted(banks.items(),
                       key=lambda kv: lex_key(kv[1], order_forward))

print("\nЛексикографический порядок (критерии по важности):")
for i, (name, _) in enumerate(banks_forward, 1):
    print(f"{i:2d}. {name}")


# Обратный порядок критериев
order_backward = list(reversed(importance))
banks_backward = sorted(banks.items(),
                        key=lambda kv: lex_key(kv[1], order_backward))

print("\nЛексикографический порядок (обратный порядок критериев):")
for i, (name, _) in enumerate(banks_backward, 1):
    print(f"{i:2d}. {name}")
```

## Пояснение

### Часть 1. Тесты

Для каждого критерия задан порог и направление:

- **min** — чем меньше значение, тем лучше (например, ставка);
- **max** — чем больше, тем лучше (например, максимальная сумма).

Банк получает 1 балл за каждый пройденный тест.  
Чем больше баллов — тем выше банк в рейтинге.

### Часть 2. Парето

Банк $A$ **доминирует** банк $B$, если:

- по всем критериям $A$ не хуже $B$;
- хотя бы по одному критерию $A$ строго лучше.

Банки, которые никто не доминирует, образуют **Парето-фронт** — это «лучшие компромиссные» варианты.

### Часть 3. Лексикографический порядок

Критерии располагаются по важности. Сравнение банков идёт **последовательно**:

1. Сначала по самому важному критерию.
2. При равенстве — по следующему.
3. И так далее.

**Прямой порядок** — от самого важного к наименее важному.  
**Обратный порядок** — от наименее важного к самому важному.
